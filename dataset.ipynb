```python
import pandas as pd
import random
import matplotlib.pyplot as plt
```

```python
# y = 0
rows = []
for i in range(15):
    x = random.randint(-50, 50)
    y =random.randint(-50, -1)
    label = 0
    rows.append((x, y, label))
for i in range(15):
    x = random.randint(-50, 50)
    y =random.randint(1, 50)
    label = 1
    rows.append((x, y, label))

data = pd.DataFrame(rows, columns=['x', 'y', 'label'])
data.to_csv('data.csv', index=False)
```

```python
data = pd.read_csv('data.csv')
class0 = data[data["label"] == 0]
class1 = data[data["label"] == 1]

plt.scatter(class0["x"], class0["y"], color="blue", label="0 klasė")
plt.scatter(class1["x"], class1["y"], color="red", label="1 klasė")

plt.legend()
plt.grid(True)
plt.show()
```
