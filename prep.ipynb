```python
import pandas as pd
data = pd.read_csv("breast-cancer-wisconsin.data", sep = ',', header = None)
data
```

```python

data.columns = [
"Sample_code_number",
"Clump_thickness",
"Uniformity_of_cell_size",
"Uniformity_of_cell_shape",
"Marginal_adhesion",
"Single_epithelial_cell_size",
"Bare_nuclei",
"Bland_chromatin",
"Normal_nucleoli",
"Mitoses",
"Class"
]

```

```python
data.rename(columns = {"Class": "label"}, inplace = True)

mask_0 = data["label"] == 2
data.loc[mask_0, "label"] = 0 #nepiktybinis
mask_1 = data["label"] == 4
data.loc[mask_1, "label"] = 1 #piktybinis
```

```python
for col in data.columns:
    mask= data[col] == '?'
    data.loc[mask, col] = None
    print(f"{col} has {data[col].isna().sum()} missing values")
```

```python
missing_val_index = []
for col in data.columns:
    missing_val_index.extend(data[data[col].isna()].index.tolist())
data.drop(index = missing_val_index, inplace = True)
data.drop(columns = ["Sample_code_number"], inplace = True)
```

```python
feature_cols = [c for c in data.columns if c != 'label']
data[feature_cols] = data[feature_cols].apply(pd.to_numeric)
```

```python
data = data.sample(frac=1).reset_index(drop=True)
```

```python
data.to_csv("breast_cancer.csv", index = False)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

data = pd.read_csv("breast_cancer.csv")


train_validation, test = train_test_split(data, test_size=0.1, random_state=67)

train, validation = train_test_split(train_validation, test_size=0.1, random_state=67)
train.to_csv("train.csv", index = False)
validation.to_csv("validation.csv", index = False)
test.to_csv("test.csv", index = False)


```
